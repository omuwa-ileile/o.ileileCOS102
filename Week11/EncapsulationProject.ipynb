```python
import csv
import pandas as pd


class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.grade = grade
        self.age = age
        self.__error_message = "Error: Student age is greater than 25."

    # Getter method for encapsulated error message
    def get_error_message(self):
        return self.__error_message

    # Method to determine category
    def classify_student(self):
        if 14 < self.age < 18:
            return "The_Pirates"
        elif 18 < self.age < 22:
            return "The_Yankees"
        elif 22 < self.age < 25:
            return "The_Bulls"
        elif self.age > 25:
            return self.get_error_message()
        else:
            return "No Category"


# Lists to store records
pirates = []
yankees = []
bulls = []

# Read data from SIS.csv
with open("SIS.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        name = row["Name"]
        age = int(row["Age"])
        grade = row["Grade"]

        student = Student(name, age, grade)
        category = student.classify_student()

        if category == "The_Pirates":
            pirates.append(row)

        elif category == "The_Yankees":
            yankees.append(row)

        elif category == "The_Bulls":
            bulls.append(row)

        else:
            print(name, "-", category)


# Create CSV files for each category
pd.DataFrame(pirates).to_csv("The_Pirates.csv", index=False)
pd.DataFrame(yankees).to_csv("The_Yankees.csv", index=False)
pd.DataFrame(bulls).to_csv("The_Bulls.csv", index=False)


# Display the contents of each category
print("\nTHE PIRATES")
print(pd.DataFrame(pirates))

print("\nTHE YANKEES")
print(pd.DataFrame(yankees))

print("\nTHE BULLS")
print(pd.DataFrame(bulls))
```
