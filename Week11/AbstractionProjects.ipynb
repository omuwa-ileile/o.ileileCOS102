```python
from abc import ABC, abstractmethod

# Abstract Base Class
class coup_de_ecriva(ABC):

    @abstractmethod
    def Fan_Page(self):
        pass


# Subclass 1
class FC_Cirok(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to FC Cirok Fan Page')


# Subclass 2
class Madiba_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to Madiba FC Fan Page')


# Subclass 3
class Blue_Jay_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to Blue Jay FC Fan Page')


# Subclass 4
class TSG_Walker(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to TSG Walker Fan Page')


# User Input
club = input("Enter the club name: ")

# Instantiate based on input
if club.lower() == "fc cirok":
    obj = FC_Cirok()
elif club.lower() == "madiba fc":
    obj = Madiba_FC()
elif club.lower() == "blue jay fc":
    obj = Blue_Jay_FC()
elif club.lower() == "tsg walker":
    obj = TSG_Walker()
else:
    print("Club not found")
    obj = None

# Call method if valid club selected
if obj:
    obj.Fan_Page()
```

```python
from abc import ABC, abstractmethod
import tkinter as tk
from tkinter import messagebox
from tkinter import ttk


# Abstract Base Class
class ExternalVendors(ABC):

    @abstractmethod
    def menu(self):
        pass


# Subclass 1
class CooperativeHostel(ExternalVendors):
    def menu(self):
        return """Cooperative Cafeteria Menu
--------------------------------
Jollof Rice and Stew  - N200
White Rice and Stew   - N200
Fried Rice            - N200
Salad                 - N100
Plantain              - N100
"""


# Subclass 2
class FaithHostel(ExternalVendors):
    def menu(self):
        return """Faith Hostel Cafeteria Menu
--------------------------------
Fried Rice            - N400
White Rice and Stew   - N400
Jollof Rice           - N400
Beans                 - N200
Chicken               - N1000
"""


# Subclass 3
class StudentCenter(ExternalVendors):
    def menu(self):
        return """Student Centre Cafeteria Menu
--------------------------------
Chicken Fried Rice    - N800
Pomo Sauce           - N300
Spaghetti Jollof     - N500
Amala/Ewedu          - N500
Semo with Eforiro    - N500
"""


# Vendor selector (FIXED using dictionary)
def get_vendor(vendor_name):
    vendor_name = vendor_name.strip().lower()

    vendors = {
        "cooperative hostel": CooperativeHostel,
        "faith hostel": FaithHostel,
        "student center": StudentCenter
    }

    vendor_class = vendors.get(vendor_name)

    return vendor_class() if vendor_class else None


# Show menu function
def show_menu():
    vendor = vendor_var.get()

    obj = get_vendor(vendor)

    output_box.delete("1.0", tk.END)

    if obj:
        output_box.insert(tk.END, obj.menu())
    else:
        messagebox.showerror("Error", "Vendor not found!")


# ---------------- GUI ----------------
window = tk.Tk()
window.title("PAU External Food Vendors System")
window.geometry("500x450")


# Label
label = tk.Label(window, text="Select Vendor Name:")
label.pack(pady=5)


# Dropdown (FIXED INPUT METHOD)
vendor_var = tk.StringVar()

dropdown = ttk.Combobox(window, textvariable=vendor_var)
dropdown['values'] = ("Cooperative Hostel", "Faith Hostel", "Student Center")
dropdown.pack(pady=5)


# Button
btn = tk.Button(window, text="Show Menu", command=show_menu)
btn.pack(pady=10)


# Output Box
output_box = tk.Text(window, height=15, width=55)
output_box.pack(pady=10)


# Run App
window.mainloop()
```
